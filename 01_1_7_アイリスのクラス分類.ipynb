## 🎯 なぜ「1.7」を徹底的にやるべきなのか？

機械学習のプロジェクトは、基本的に以下のステップで進みます。1.7のセクションでは、この一連の流れをわずか数ページのコードで体験できます。
- データの確認：アヤメのデータ（特徴量とラベル）がどうなっているかを見る
- データの分割：データを「訓練用（勉強用）」と「テスト用（本番の試験用）」に分ける
- データの観察：散布図（ペアプロット）を描いて、データの特徴を目で見る
- モデルの構築：k-近傍法（k-NN）というアルゴリズムを使って学習させる
- 予測と評価：新しいアヤメの予測を行い、テストデータを使って「正解率」を出す

<br>
後半の章（2章や3章など）に進むと、使うアルゴリズム（道具）が変わるだけで、「分割して、学習させて、評価する」という全体の流れは1.7のセクションと全く同じです。<br>そのため、ここをクリアしておくと、2章以降の読み進めやすさが段違いに上がります。
