## 🌊 今回やった「k-近傍法」の場合（鉄板をたたかない）

k-近傍法は、鉄板を引き伸ばしたり叩いたりしません。<br><br>
* イメージ
  * フラットな鉄板の上に、112個の訓練データ（赤・緑・青の杭）をそのままグサグサと刺します。
* 境界線
  * それぞれの杭から「一番近いエリア」ごとに色を塗っていきます。
  * できあがった境界線は、ジグザグとしたモザイク模様のようになります。
* 分類と回帰
  * 上から新しいアヤメのピンを落としたとき、それが「どの色の杭のエリアに落ちたか（分類）」、あるいは「周りの杭の高さの平均はいくつか（回帰）」で判定します。


## 📐 第2章で学ぶ「線形モデル（ロジスティック回帰など）」の場合（鉄板を曲げない）

線形モデルは、ディープラーニングの「活性化関数（トントン叩く工程）」を取り除いた、最もシンプルな状態です。<br><br>
* イメージ
  * 鉄板は真っ直ぐ（平ら）なままです。
  * 重みで傾け（回転）、バイアスで高さ（平行移動）を調整します。
* 境界線
  * 鉄板を斜めに傾けて、一定の高さに水を注ぐと、水面と鉄板が交わる線は「真っ直ぐな直線（または平面）」になります。
  * これが分類の境界線です。データを直線だけでスパッと一刀両断するイメージです。
* 回帰
  * 水を注がずに、傾いた鉄板のその位置の「リアルな高さ」をそのまま予測値とします（これが線形回帰です）。
<br><br>
## 🌳 第2章で学ぶ「決定木（デシジョンツリー）」の場合（鉄板を階段状にする）

決定木は、Yes/Noのクイズを繰り返すアルゴリズムです。<br><br>
* イメージ
  * 鉄板を引き伸ばすのではなく、軸に対して垂直に「パキン、パキン」と折り曲げて、階段（ブロック）のような凸凹を作っていきます。
* 境界線
  * 水を注ぐと、ブロックの段差に沿って、カクカクとした四角い境界線が生まれます。
* 回帰
  * 新しいピンが落ちた場所の「ブロックの段差の高さ」をそのまま予測値とします。
<br><br>
